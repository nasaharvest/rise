# Validate ukrop Format

Checks a standardized output file (from `preprocess_data.ipynb`) against
ukrop's actual ingestion contract - required fields, enum values, the
single-date-per-operation rule, etc. This answers **"will ukrop accept this
file"**, not "is this good data" (that's `qa_checks.ipynb`'s job).

## Expected Input
- A standardized geometry file, e.g. `data/check_file.geojson`, output of
  `preprocess_data.ipynb`.

## Output
- A printed report: PASS/FAIL plus a list of specific issues (each one
  traceable to a rule in `CLAUDE.md`'s ukrop contract section) and
  informational notes. Nothing is written to disk - fix issues in
  `preprocess_data.ipynb` and re-export, then re-run this notebook.
